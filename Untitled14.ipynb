```python
import pandas as pd
import folium
import requests
import json

# 카카오 맵 API 키를 여기에 입력하세요!
KAKAO_API_KEY = "6c37421c71c57b25c6ab941553fb4371"

def addr_lat_lon(addr):
    url = f'https://dapi.kakao.com/v2/local/search/address.json?query={addr}'
    headers = {"Authorization": f"KakaoAK {KAKAO_API_KEY}"}
    try:
        response = requests.get(url, headers=headers)
        result = json.loads(response.text)
        print(f"응답성공")
        #print(f"API 응답 for '{addr}': {result}") # 응답 내용 출력 (디버깅용)
        if result['documents']:
            match_first = result['documents'][0]['address']
            return float(match_first['x']), float(match_first['y']) # 경도(x), 위도(y) 순서로 반환
        else:
            return None, None
    except Exception as e:
        print(f"Error during geocoding for '{addr}': {e}")
        return None, None


# 서울 중심부 위도, 경도 (카카오맵 기준)
seoul_center = [37.5718, 126.9858]

# 초기 확대 수준 설정 zoom_start
# 지도 생성 (서울 중심으로 초기화)
m = folium.Map(location=seoul_center, zoom_start=18)

# 데이터 파일 이름 (본인의 파일명으로 변경해주세요)
data_file = "C:\\Users\\USER\\Desktop\\python\\맛집목록.xlsx"  # 또는 "맛집목록.csv"

try:
    # 엑셀 파일 읽기
    if data_file.endswith(".xlsx"):
        df = pd.read_excel(data_file)
    # CSV 파일 읽기
    elif data_file.endswith(".csv"):
        df = pd.read_csv(data_file, encoding='utf-8')
    else:
        raise ValueError("지원하지 않는 파일 형식입니다. (.xlsx 또는 .csv)")

    # '상호명'과 '위치' 컬럼이 있는지 확인
    if '상호명' not in df.columns or '위치' not in df.columns:
        print("오류: 엑셀/CSV 파일에 '상호명'과 '위치' 컬럼이 있어야 합니다.")
    else:
        for index, row in df.iterrows():
            name = row['상호명']
            location_str = row['위치']
            memo = row.get('메모', '')  # '메모' 컬럼이 없으면 빈 문자열 사용

            longitude, latitude = addr_lat_lon(location_str)

            if latitude is not None and longitude is not None:
                # 팝업 HTML 콘텐츠 생성 및 스타일 적용
                popup_html = f"""
                    <div style="width: 300px; font-size: 14px;">
                        <b>{name}</b><br>
                        위치: {location_str}<br>
                        메모: {memo}
                    </div>
                """
                popup = folium.Popup(popup_html, max_width=300)
                folium.Marker([latitude, longitude], popup=popup).add_to(m)
            else:
                print(f"'{name}' ({location_str}) 주소의 위도, 경도를 찾을 수 없습니다.")

        # 지도를 HTML 파일로 저장
        m.save("서울맛집지도_카카오_API.html")
        print("서울맛집지도_카카오_API.html 파일이 생성되었습니다.")

except FileNotFoundError:
    print(f"오류: '{data_file}' 파일을 찾을 수 없습니다.")
except ValueError as e:
    print(e)
except Exception as e:
    print(f"예상치 못한 오류 발생: {e}")
```
